```python
missing_counts = None

for chunk in pd.read_csv(
    output_path,
    chunksize=500000
):
    chunk_missing = chunk.isnull().sum()

    if missing_counts is None:
        missing_counts = chunk_missing
    else:
        missing_counts = missing_counts.add(chunk_missing, fill_value=0)

missing_counts.astype("int64")
```

```python
missing_percent = (
    missing_counts / 5442964 * 100
).round(2)

missing_percent
```

```python
missing_summary = pd.DataFrame({
    "missing_count": missing_counts.astype("int64"),
    "missing_percent": missing_percent
}).sort_values("missing_percent", ascending=False)

missing_summary
```

```python
seen_ids = set()
duplicate_id_count = 0

for chunk in pd.read_csv(
    output_path,
    usecols=["Complaint ID"],
    chunksize=500000
):
    for complaint_id in chunk["Complaint ID"]:
        if complaint_id in seen_ids:
            duplicate_id_count += 1
        else:
            seen_ids.add(complaint_id)

duplicate_id_count
```

```python
zip_format_counts = {
    "5-digit": 0,
    "masked": 0,
    "missing": 0,
    "other": 0
}

for zipcode, count in zip_counts.items():

    if pd.isna(zipcode):
        zip_format_counts["missing"] += count

    elif str(zipcode).isdigit() and len(str(zipcode)) == 5:
        zip_format_counts["5-digit"] += count

    elif "X" in str(zipcode).upper():
        zip_format_counts["masked"] += count

    else:
        zip_format_counts["other"] += count

zip_format_counts
```

```python
missing_counts = None
total_rows = 0

for chunk in pd.read_csv(
    output_path,
    chunksize=500000
):
    total_rows += len(chunk)

    chunk_missing = chunk.isna().sum()

    if missing_counts is None:
        missing_counts = chunk_missing

    else:
        missing_counts = missing_counts.add(chunk_missing, fill_value=0)

missing_summary = pd.DataFrame({
    "missing_count": missing_counts.astype(int),
    "missing_percent": (missing_counts / total_rows * 100).round(2)
})

missing_summary.sort_values(
    "missing_percent",
    ascending=False
)
```

```python
missing_subissue_by_product = {}

for chunk in pd.read_csv(
    output_path,
    usecols=["Product", "Sub-issue"],
    chunksize=500000
):
    missing_rows = chunk[chunk["Sub-issue"].isna()]

    counts = missing_rows["Product"].value_counts()

    for product, count in counts.items():
        missing_subissue_by_product[product] = (missing_subissue_by_product.get(product, 0) + count)

missing_subissue_by_product = pd.Series(missing_subissue_by_product).sort_values(ascending=False)

missing_subissue_by_product
```

```python
missing_state_zip_counts = {}

for chunk in pd.read_csv(
    output_path, 
    usecols=["State", "ZIP code"],
    chunksize=500000
):
    missing_state_rows = chunk[chunk["State"].isna()]

    counts = missing_state_rows["ZIP code"].value_counts(dropna=False)

    for zipcode, count in counts.items():
        missing_state_zip_counts[zipcode] = (missing_state_zip_counts.get(zipcode, 0) + count)

missing_state_zip_summary = pd.Series(missing_state_zip_counts).sort_values(ascending=False)

print("Total complaints with missing State:", missing_state_zip_summary.sum())
print("Number of ZIP values among missingState records", len(missing_state_zip_summary))

missing_state_zip_summary.head(30)
```

```python
missing_state_counts = {}

for chunk in pd.read_csv(
    output_path, 
    usecols=["ZIP code", "State"],
    chunksize=500000
):
    missing_state_rows = chunk[chunk["State"].notna()]

    counts = missing_state_rows.groupby(["ZIP code", "State"]).size()

    for (zipcode, state), count in counts.items():
        key = (zipcode, state)
        
        missing_state_counts[key] = (missing_state_counts.get(key, 0) + count)

print("Number of ZIP-State combinations:", len(missing_state_counts))
```
